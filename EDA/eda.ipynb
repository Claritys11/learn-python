## Latihan 1: Operasi Dasar NumPy Array

```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

```python
# Coba kalau pakai list Python biasa
harga_list = [5000, 7000, 3000]
try:
    print(harga_list * 0.9)
except TypeError as e:
    print('Error:', e)

# List dikali bilangan bulat malah diulang, bukan dihitung
print('harga_list * 2 =', harga_list * 2)

# Kalau tetap mau pakai list, harus diloop satu-satu
print('Pakai list comprehension:', [h * 0.9 for h in harga_list])
```

**Tugas Analisis 1:**

Di NumPy, `harga * 0.9` mengalikan tiap elemen, jadi semua harga kena diskon 10%. Kalau pakai list, `[5000,7000,3000] * 0.9` error (`TypeError`), karena `*` pada list artinya mengulang isi list, bukan mengalikan angka.

## Latihan 2: Membuat Series & DataFrame

```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

**Tugas Analisis 2:**

Kolom yang kosong: `menu` (index 4) dan `terjual` (Mie Ayam, index 2). Kalau langsung dianalisis, hasilnya bisa salah atau error karena ada data yang hilang.

## Latihan 3: Data Loading & Inspection

```python
df = pd.read_csv('data_kantin.csv')

print(df.head())        # 5 baris pertama
print(df.info())        # tipe data & jumlah non-null tiap kolom
print(df.describe())    # statistik ringkas kolom numerik
print(df.shape)         # jumlah (baris, kolom)
```

**Tugas Analisis 3:**

Dari 110 baris, `menu` cuma 107 non-null (3 kosong) dan `terjual` 104 non-null (6 kosong). Artinya dua kolom itu punya missing value yang harus dibersihkan dulu.

## Latihan 4: Menangani Missing Value

```python
print(df.isnull().sum())              # jumlah data kosong tiap kolom

df['terjual'] = df['terjual'].fillna(0)           # isi kekosongan dengan 0
df = df.dropna(subset=['menu'])                    # hapus baris jika kolom menu kosong

print('\nSetelah dibersihkan:')
print(df.isnull().sum())
print('Jumlah baris sekarang:', len(df))
```

**Tugas Analisis 4:**

`terjual` diisi 0 karena kosong bisa dianggap nggak ada yang laku, dan barisnya masih berguna. Baris yang `menu`-nya kosong dihapus karena kita nggak bisa nebak itu menu apa, jadi datanya nggak kepakai buat analisis per menu.

## Latihan 5: Menangani Duplikat dan Tipe Data

```python
print('Jumlah baris sebelum:', len(df))
print(df.duplicated().sum())              # jumlah baris duplikat
df = df.drop_duplicates()
print('Jumlah baris sesudah:', len(df))

df['harga'] = df['harga'].astype(int)           # memastikan tipe data harga adalah integer
print(df.dtypes)
```

**Tugas Analisis 5:**

Ada 5 duplikat, jadi barisnya berkurang dari 107 ke 102 (dari awal: 110 → 107 → 102). Tipe data harus benar supaya perhitungan dan sorting jalan dengan benar. Misalnya, kalau harga masih berupa teks, nggak bisa dikalikan.

## Latihan 6: Data Manipulation (Filtering, Sorting, Groupby)

```python
laris = df[df['terjual'] > 20]                      # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual']                   # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum()                     # agregasi
print(ringkasan)
```

```python
# Biar lebih gampang dibaca: urutkan dari yang paling besar,
# terus bandingkan sama jumlah porsi yang terjual
perbandingan = df.groupby('menu').agg(
    porsi_terjual=('terjual', 'sum'),
    total_pendapatan=('total_pendapatan', 'sum')
).sort_values('total_pendapatan', ascending=False)
print(perbandingan)
print('\nJumlah transaksi dengan terjual > 20:', len(laris))
```

**Tugas Analisis 6:**

Pendapatan tertinggi ada di Nasi Goreng (Rp3.924.000). Info ini bisa dipakai kantin buat mengutamakan stok bahan Nasi Goreng dan bikin strategi jual untuk menu yang pendapatannya kecil.
