```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

**Latihan 1 : **Pada NumPy array, harga * 0.9 akan mengalikan setiap elemen array dengan 0.9 sehingga harga masing-masing barang menjadi 90% dari harga awal. Sedangkan jika menggunakan list Python biasa seperti [5000,7000,3000] * 0.9, akan terjadi TypeError karena list Python tidak mendukung perkalian dengan bilangan koma secara element wise. NumPy array dirancang untuk melakukan operasi matematika secara langsung pada setiap elemen, sehingga lebih praktis untuk pengolahan dan analisis data numerik. teks yang dimiringkan

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

**Latihan 2 : **Kolom yang memiliki data kosong adalah kolom menu dan terjual. Jika data tersebut langsung dianalisis tanpa dibersihkan, hasil analisis dapat menjadi tidak akurat karena terdapat informasi yang hilang. Data kosong juga dapat menyebabkan kesalahan dalam perhitungan atau mengganggu proses analisis dan pembuatan model machine learning.

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

**Latihan 3 : **Dari hasil df.info(), kolom menu dan terjual memiliki jumlah non-null yang lebih sedikit dari jumlah baris total. Kolom menu memiliki 107 data dari 110 baris sehingga terdapat 3 data kosong, sedangkan kolom terjual memiliki 104 data sehingga terdapat 6 data kosong. Artinya, kedua kolom tersebut memiliki missing value yang perlu dibersihkan sebelum dilakukan analisis agar hasilnya lebih akurat.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

**Latihan 4 : **Kolom terjual diisi dengan 0 karena nilai kosong masih dapat diasumsikan sebagai tidak ada barang yang terjual, sehingga data tetap dapat digunakan dalam perhitungan. Sedangkan baris dengan menu kosong dihapus karena menu merupakan informasi penting untuk mengidentifikasi barang. Jika nama menu tidak diketahui, data tersebut sulit digunakan untuk analisis berdasarkan menu. Namun, pengisian terjual dengan 0 harus dilakukan hanya jika kosong memang memiliki arti "tidak terjual", bukan sekadar data yang belum diketahui.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

**Latihan 5 : **Setelah drop_duplicates(), jumlah baris dataset menjadi 105 baris, dari sebelumnya 110 baris, karena terdapat 5 baris duplikat. Memastikan tipe data (dtypes) sudah benar penting agar setiap kolom dapat digunakan sesuai fungsinya. Misalnya, kolom harga harus berupa integer agar dapat digunakan dengan tepat dalam perhitungan statistik dan operasi matematika.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

**Latihan 6 : **Berdasarkan hasil groupby(), menu yang menghasilkan total_pendapatan tertinggi adalah Nasi Goreng dengan total Rp3.924.000. Informasi ini dapat membantu kantin sekolah menentukan menu yang perlu diprioritaskan, terutama dalam menyediakan stok bahan baku dan menentukan strategi penjualan. Kantin juga dapat membandingkan pendapatan setiap menu untuk mengetahui menu mana yang memberikan kontribusi besar terhadap pendapatan.
